# HTTP Request nədir?

HTTP Request bir proqramın internet üzərindən başqa bir serverə **məlumat göndərməsi və ya məlumat istəməsi** üsuludur.

Sadə dillə desək, proqramımız başqa bir proqramla internet üzərindən danışmaq istəyəndə HTTP Request göndərir.

Bunu belə təsəvvür edə bilərik:

```text
Sənin Python kodun
       ↓
HTTP Request
       ↓
Server / API
       ↓
HTTP Response
       ↓
Sənin Python kodun
```

Məsələn, Python proqramımız AI modelindən cavab almaq istəyir:

```text
Python proqramı
       ↓
"Mənə Python-u izah et"
       ↓
HTTP Request
       ↓
AI API
       ↓
AI Model
       ↓
HTTP Response
       ↓
Python proqramı
```

---

## HTTP Request-in əsas hissələri

Bir HTTP Request əsasən aşağıdakı hissələrdən ibarət ola bilər:

```text
HTTP Request
│
├── Method
├── URL
├── Headers
└── Body
```

### 1. Method

Serverə **nə etmək istədiyimizi** bildirir.

Ən çox istifadə olunan HTTP method-lar:

- `GET`
- `POST`
- `PUT`
- `PATCH`
- `DELETE`

Bu dərsdə əsas olaraq 4 növə baxacağıq.

---

## GET

Serverdən **məlumat almaq** üçün istifadə olunur.

```text
GET /users
```

Məsələn:

> "Mənə istifadəçilərin siyahısını ver."

---

## POST

Serverə **məlumat göndərmək** və ya bir əməliyyat başlatmaq üçün istifadə olunur.

```text
POST /users
```

Məsələn:

> "Bu yeni istifadəçini yarat."

AI API-lərinə request göndərərkən də çox vaxt `POST` istifadə olunur.

---

## PUT

Mövcud məlumatı **tam şəkildə yeniləmək** üçün istifadə olunur.

```text
PUT /users/123
```

Məsələn:

> "123 ID-li istifadəçinin məlumatlarını bu yeni məlumatlarla əvəz et."

---

## DELETE

Məlumatı **silmək** üçün istifadə olunur.

```text
DELETE /users/123
```

Məsələn:

> "123 ID-li istifadəçini sil."

---

## 2. URL

Request-in hara göndəriləcəyini göstərir.

Məsələn:

```text
https://api.example.com/users
```

Burada `/users` müəyyən endpoint-dir.

---

## 3. Headers

Request haqqında əlavə məlumat verir.

Məsələn:

```text
Authorization: Bearer API_KEY
Content-Type: application/json
```

`Authorization` authentication üçün istifadə oluna bilər.

`Content-Type` isə göndərilən məlumatın formatını bildirir.

---

## 4. Body

Serverə göndərdiyimiz əsas məlumatdır.

Məsələn:

```json
{
  "name": "Sita",
  "age": 27
}
```

AI API-də body daxilində model adı, prompt və digər parametrlər ola bilər.

---

# HTTP Response nədir?

Server Request-i qəbul edib emal etdikdən sonra bizə HTTP Response göndərir.

```text
Python
  ↓
HTTP Request
  ↓
API
  ↓
HTTP Response
  ↓
Python
```

Response daxilində status code, headers və body kimi məlumatlar ola bilər.

Məsələn:

```text
200 OK
```

request-in uğurlu olduğunu göstərə bilər.

---

# HTTP-ni niyə bilməliyik?

AI API ilə işləyəndə arxa planda əslində bu proses baş verir:

```text
Python
  ↓
HTTP Request
  ↓
AI API
  ↓
AI Model
  ↓
HTTP Response
  ↓
Python
```

SDK istifadə etdikdə HTTP request-i əllə yazmırıq. SDK bu işi bizim üçün sadələşdirir.

Ona görə HTTP-ni başa düşmək vacibdir.

Əks halda `endpoint`, `header`, `authentication`, `body`, `status code` kimi anlayışlar qarışıq görünə bilər.

> **HTTP Request proqramların internet üzərindən bir-biri ilə məlumat mübadiləsi etməsinin əsas mexanizmlərindən biridir.**

# API nədir?

**API — Application Programming Interface** deməkdir.

Sadə dillə desək, API iki proqramın **bir-biri ilə danışması üçün müəyyən edilmiş qaydalar və interfeysdir**.

Məsələn, bizim Python proqramımız AI modelindən istifadə etmək istəyir.

Modelin özünü kompüterimizə yükləmək əvəzinə API vasitəsilə AI serverinə request göndərə bilərik:

```text
Python Application
       ↓
       API
       ↓
AI Server
       ↓
AI Model
       ↓
   Response
```

API bizə deyir:

- hansı endpoint-ə request göndərməliyik;
- hansı HTTP method istifadə etməliyik;
- hansı məlumatları göndərə bilərik;
- authentication necə edilir;
- request hansı formatda olmalıdır;
- response hansı formatda gələcək.

---

# API-ni restoran kimi düşünək

API-ni restoranın ofisiantı kimi düşünə bilərik.

```text
Sən
 ↓
Ofisiant (API)
 ↓
Mətbəx (Server / Model)
 ↓
Ofisiant
 ↓
Sən
```

Sən mətbəxə girib yeməyi özün hazırlamırsan.

API də eyni məntiqlə işləyir.

Sən:

> "Bu prompt ilə modeldən cavab istəyirəm."

deyirsən.

API request-i serverə çatdırır və nəticəni sənə qaytarır.

---

# API necə işləyir?

Məsələn:

```text
1. Python proqramı request hazırlayır
             ↓
2. Request API-yə göndərilir
             ↓
3. API request-i serverə çatdırır
             ↓
4. Server request-i emal edir
             ↓
5. Model nəticə yaradır
             ↓
6. API response-u Python-a qaytarır
```

# API yalnız AI üçün istifadə olunmur

API-lər çox fərqli sistemlər arasında əlaqə yarada bilər.

Məsələn:

```text
Weather App
     ↓
Weather API

Payment App
     ↓
Payment API

Maps App
     ↓
Maps API

AI Application
     ↓
AI API
```

# AI API nədir?

AI API proqramımızın AI modelindən proqram vasitəsilə istifadə etməsinə imkan verir.

Məsələn:

```python
response = client.responses.create(
    model="gpt-5.5",
    input="Explain Python."
)
```

Burada Python proqramımız AI API ilə əlaqə yaradır.

Biz modelə birbaşa daxil olmuruq.

```text
Python
   ↓
AI API
   ↓
AI Model
   ↓
Response
```

# API-ni niyə bilməliyik?

AI developer üçün API anlayışı fundamental mövzulardan biridir.

Çünki OpenAI, Gemini, Grok, OpenRouter və digər AI platformaları ilə işləyərkən API anlayışı daim qarşımıza çıxır.

API-ni başa düşəndə aşağıdakı anlayışlar daha rahat olur:

- endpoint
- authentication
- API key
- request
- response
- parameters
- headers
- body
- status codes

Ən vacib fikir:

> **API proqramların bir-biri ilə danışması üçün müəyyən edilmiş qaydalar və interfeysdir.**

HTTP isə bu əlaqənin qurulmasında istifadə olunan əsas protokollardan biridir.

```text
HTTP
 ↓
Request / Response mexanizmi

API
 ↓
Proqramların hansı qaydalarla danışacağını müəyyən edir
```
